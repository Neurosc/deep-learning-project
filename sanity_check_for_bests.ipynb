```python
import os
import importlib.util
import numpy as np

ROOT = os.path.expanduser("~/things_eeg")

# ------------------------------------------------------------
# Load functions from 03_run_alignment.py
# ------------------------------------------------------------

alignment_path = os.path.join(
    ROOT,
    "03_run_alignment.py"
)

spec = importlib.util.spec_from_file_location(
    "alignment",
    alignment_path,
)

alignment = importlib.util.module_from_spec(spec)
spec.loader.exec_module(alignment)


# ------------------------------------------------------------
# Load already prepared EEG
# ------------------------------------------------------------

train_eeg = np.load(
    os.path.join(
        ROOT,
        "eeg_prepared",
        "sub-01_train_avg.npy",
    )
)

test_eeg = np.load(
    os.path.join(
        ROOT,
        "eeg_prepared",
        "sub-01_test_avg.npy",
    )
)


# ------------------------------------------------------------
# Layers to test
# ------------------------------------------------------------

TARGETS = [
    "RN50__layer3",
    "ViT-B-16__block6",
    "ViT-L-14__block8",
    "ViT-L-14__block24",
]


# ------------------------------------------------------------
# Run only full 1-second sanity checks
# ------------------------------------------------------------

for target in TARGETS:

    print(
        f"\n=== Full 1-second sanity check: {target} ==="
    )

    alignment.sanity_check(
        train_eeg,
        test_eeg,
        target,
        seed=0,
    )
```
